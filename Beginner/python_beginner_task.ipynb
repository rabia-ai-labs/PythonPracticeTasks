````python
# Task 4
total_steps = 0

for day in range(1,8):
  daily_steps = int(input(f"Enter steps of Day:{day} "))

  total_steps = total_steps + daily_steps

daily_average = total_steps / 7

if daily_average >=5000:
  status = "Active Week"
else:
  status = "Needs More Activity"

print(f"\nTotal Steps: {total_steps}")
print("Daily Average:", int(daily_average))
print(f"Status: {status}")
````

````python
# Task 5

secret_pin = 1234
pin = int(input("Please Enter your PIN: "))

for i in range(1, 4):
    if secret_pin == pin:
        print("Access Granted.")
        break
    else:
        print("Access Denied.")
        print(f"Attempt {i} failed")

        if i == 3:
            print("Account Blocked")
            break

        pin = int(input("Please Enter your PIN: "))
````
