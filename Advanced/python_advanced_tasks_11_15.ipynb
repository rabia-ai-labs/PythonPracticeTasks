````python
# TASK 11: Personal Expense Tracker

budget = float(input("Enter your monthly budget: Rs. "))

while budget < 0:
    print("Budget cannot be negative.")
    budget = float(input("Enter your monthly budget: Rs. "))

num_expenses = int(input("How many expenses do you want to record? "))

while num_expenses < 0:
    print("Number of expenses cannot be negative.")
    num_expenses = int(input("How many expenses do you want to record? "))

expenses_list = []

categories = ["Food", "Transport", "Shopping", "Bills", "Other"]

for i in range(num_expenses):
    print(f"\n--- Expense {i + 1} ---")

    name = input("Enter expense name (e.g., Biryani, Petrol): ").strip()

    while not name:
        print("Expense name cannot be empty.")
        name = input("Enter expense name: ").strip()

    while True:
        category = input(
            "Enter category (Food, Transport, Shopping, Bills, Other): "
        ).strip().title()

        if category in categories:
            break
        else:
            print("Invalid category. Please enter a valid category.")

    amount = float(input("Enter amount: Rs. "))

    while amount < 0:
        print("Amount cannot be negative.")
        amount = float(input("Enter amount: Rs. "))

    expense_dict = {
        "name": name,
        "category": category,
        "amount": amount
    }

    expenses_list.append(expense_dict)


def generate_report(all_expenses, user_budget):
    print("\n========================================")
    print("          FINAL EXPENSE REPORT          ")
    print("========================================")

    total_spending = 0

    category_totals = {
        "Food": 0,
        "Transport": 0,
        "Shopping": 0,
        "Bills": 0,
        "Other": 0
    }

    highest_expense = None

    for item in all_expenses:
        total_spending += item["amount"]

        category_totals[item["category"]] += item["amount"]

        if highest_expense is None or item["amount"] > highest_expense["amount"]:
            highest_expense = item

    print(f"Total Budget: Rs. {user_budget:.2f}")
    print(f"Total Spent:  Rs. {total_spending:.2f}")
    print("----------------------------------------")
    print("Category Wise Spending:")

    for category, amount in category_totals.items():
        print(f" - {category}: Rs. {amount:.2f}")

    print("----------------------------------------")

    if highest_expense is not None:
        print(
            f"Highest Expense: {highest_expense['name']} "
            f"(Rs. {highest_expense['amount']:.2f})"
        )
    else:
        print("No expenses recorded.")

    print("----------------------------------------")

    if total_spending > user_budget:
        print("WARNING: You have exceeded your budget!")
    else:
        print(f"Remaining Budget: Rs. {user_budget - total_spending:.2f}")
        print("Safe Zone: You are within your budget.")

    print("========================================")


generate_report(expenses_list, budget)
````

````python
# Task 12: Store Inventory Manager
inventory = {}

def add_product(name, price, quantity):
    if name in inventory:
        print("Product already exists! Use Restock option instead.")
    else:
        inventory[name] = {"price": price, "quantity": quantity}
        print(f"Product '{name}' added successfully.")


def sell_product(name, quantity):
    if name in inventory:
        current_stock = inventory[name]["quantity"]

        if current_stock >= quantity:
            inventory[name]["quantity"] -= quantity
            print(f"Sold {quantity} units of '{name}'.")

            if inventory[name]["quantity"] < 5:
                print(f"Warning: '{name}' is running low on stock! ({inventory[name]['quantity']} remaining)")
        else:
            print(f"Insufficient Stock! Only {current_stock} units available.")
    else:
        print("Product not found in inventory.")


def restock_product(name, quantity):
    if name in inventory:
        inventory[name]["quantity"] += quantity
        print(f"Restocked '{name}'. New quantity: {inventory[name]['quantity']}")
    else:
        print("Product not found. Add it as a new product first.")


def search_product(name):
    if name in inventory:
        print(f"Found: {name} | Price: Rs. {inventory[name]['price']} | Stock: {inventory[name]['quantity']}")
    else:
        print("Product not found.")


def display_inventory():
    if not inventory:
        print("Inventory is empty.")
        return

    print("\n--- Current Inventory Status ---")
    total_inventory_value = 0
    low_stock_products = []

    for name, details in inventory.items():
        prod_value = details["price"] * details["quantity"]
        total_inventory_value += prod_value

        print(f"{name} -> Price: Rs. {details['price']} | Stock: {details['quantity']} | Total Value: Rs. {prod_value}")

        if details["quantity"] < 5:
            low_stock_products.append(name)

    print("--------------------------------")
    print(f"Total Shop/Inventory Value: Rs. {total_inventory_value}")

    if low_stock_products:
        print(f"Low-Stock Alerts (Below 5 units): {', '.join(low_stock_products)}")

    print("--------------------------------")


while True:
    print("\n=== Shop Inventory Menu ===")
    print("1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Exit")

    choice = input("\nEnter choice (1-6): ")

    if choice == "1":
        name = input("Enter product name: ")
        price = float(input("Enter price: Rs. "))
        qty = int(input("Enter starting quantity: "))
        add_product(name, price, qty)

    elif choice == "2":
        name = input("Enter product name to sell: ")
        qty = int(input("Enter quantity to sell: "))
        sell_product(name, qty)

    elif choice == "3":
        name = input("Enter product name to restock: ")
        qty = int(input("Enter quantity to add: "))
        restock_product(name, qty)

    elif choice == "4":
        name = input("Enter product name to search: ")
        search_product(name)

    elif choice == "5":
        display_inventory()

    elif choice == "6":
        print("Goodbye! Closing Inventory Manager.")
        break

    else:
        print("Invalid choice! Enter a number between 1 and 6.")
````

````python
# Task 13: Library Book Management System

library_books = [
    {"id": "101", "title": "Python Basics", "author": "John", "available": True},
    {"id": "102", "title": "Data Science 101", "author": "Sara", "available": True}
]

def add_book():
    book_id = input("Enter Unique Book ID: ")

    for book in library_books:
        if book["id"] == book_id:
            print("Error: Book ID already exists! ID must be unique.")
            return

    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    new_book = {"id": book_id, "title": title, "author": author, "available": True}
    library_books.append(new_book)
    print(f"Book '{title}' added successfully.")


def display_books():
    if not library_books:
        print("Library is empty.")
        return

    print("\n--- Library Catalog ---")
    for book in library_books:
        status = "Available" if book["available"] else "Borrowed (Not Available)"
        print(f"ID: {book['id']} | Title: {book['title']} | Author: {book['author']} | Status: {status}")


def search_book():
    search_title = input("Enter book title to search: ").lower()
    found = False

    for book in library_books:
        if search_title in book["title"].lower():
            status = "Available" if book["available"] else "Borrowed"
            print(f"Found -> ID: {book['id']} | Title: {book['title']} | Status: {status}")
            found = True

    if not found:
        print("No book found with this title.")


def borrow_book():
    book_id = input("Enter Book ID to borrow: ")

    for book in library_books:
        if book["id"] == book_id:
            if book["available"]:
                book["available"] = False
                print(f"Success! You have borrowed '{book['title']}'.")
                return
            else:
                print("Sorry, this book is already borrowed by someone else.")
                return

    print("Invalid Book ID! Book not found.")


def return_book():
    book_id = input("Enter Book ID to return: ")

    for book in library_books:
        if book["id"] == book_id:
            if not book["available"]:
                book["available"] = True
                print(f"Thank you! '{book['title']}' has been returned and is now available.")
                return
            else:
                print("This book was never borrowed; it's already in the library.")
                return

    print("Invalid Book ID! Book not found.")


while True:
    print("\n=== Library Management Menu ===")
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = input("\nEnter choice (1-6): ")

    if choice == "1":
        add_book()
    elif choice == "2":
        display_books()
    elif choice == "3":
        search_book()
    elif choice == "4":
        borrow_book()
    elif choice == "5":
        return_book()
    elif choice == "6":
        print("Exiting Library System. Goodbye!")
        break
    else:
        print("Invalid choice! Please enter a number between 1 and 6.")
````

````python
# Task 14: Restaurant Ordering System

menu = {
    "Burger": 500,
    "Pizza": 1200,
    "Pasta": 800,
    "Drink": 200
}

order_list = []

while True:
    print("\n=== RESTAURANT MENU ===")
    for item, price in menu.items():
        print(f" - {item}: Rs. {price}")
    print("==========================")

    item_name = input("\nEnter the item name you want to order: ").capitalize()

    if item_name in menu:
        quantity = int(input(f"Enter quantity for {item_name}: "))

        price = menu[item_name]
        subtotal = price * quantity

        item_order = {
            "item": item_name,
            "price": price,
            "quantity": quantity,
            "subtotal": subtotal
        }

        order_list.append(item_order)
        print(f"Added {quantity} x {item_name} to your cart.")
    else:
        print("Sorry, this item is not available in our menu.")

    more = input("\nDo you want to order more items? (yes/no): ").lower()
    if more == "no":
        break

def generate_receipt(cart):
    if not cart:
        print("\nYour cart is empty. No receipt generated.")
        return

    print("\n========================================")
    print("          FINAL RECEIPT          ")
    print("========================================")

    total_bill = 0

    for order in cart:
        print(f"{order['item']} x {order['quantity']} | Rate: Rs. {order['price']} | Subtotal: Rs. {order['subtotal']}")
        total_bill = total_bill + order["subtotal"]

    tax = total_bill * 13 / 100
    bill_after_tax = total_bill + tax

    print("----------------------------------------")
    print(f"Items Total:      Rs. {total_bill}")
    print(f"GST Tax (13%):    Rs. {tax}")
    print(f"Subtotal + Tax:   Rs. {bill_after_tax}")

    if total_bill >= 3000:
        discount = total_bill * 10 / 100
        final_payable = bill_after_tax - discount
        print(f"Discount (10%): Rs. {discount}")
    else:
        final_payable = bill_after_tax
        print("Discount (10%):   Rs. 0 (Bill is below Rs. 3,000)")

    print("----------------------------------------")
    print(f"FINAL PAYABLE BILL: Rs. {final_payable}")
    print("========================================")
    print("       Thank you for ordering!       ")

generate_receipt(order_list)
````

````python
# Task 15: Clinic Appointment Manager

patients = []

def add_patient():
    patient_id = int(input("Enter your id: "))
    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient already exists")
            return

    patient_name = input("Enter patient name: ")
    patient_age = int(input("Enter patient age: "))
    problem = input("Enter patient problem: ")
    time = input("Enter appointment time: ")
    emergency = input("Is emergency patient? yes/no: ")

    patients.append({
        "id": patient_id,
        "name": patient_name,
        "age": patient_age,
        "problem": problem,
        "time": time,
        "emergency": emergency
    })

    print("Patient added successfully")


def search_patient():
    patient_id = int(input("Enter patient id: "))

    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient found")
            print(f'Patient name: {patient["name"]}')
            print(f'Patient age: {patient["age"]}')
            print(f'Patient problem: {patient["problem"]}')
            print(f'Patient time: {patient["time"]}')
            print(f'Patient emergency: {patient["emergency"]}')
            return

    print("Patient not found")


def display_appointment():
    print("========== Emergency patient ==========")
    for patient in patients:
        if patient["emergency"] == "yes":
            print(f'Patient name: {patient["name"]}')
            print(f'Patient age: {patient["age"]}')
            print(f'Patient problem: {patient["problem"]}')
            print(f'Patient time: {patient["time"]}')

    print("========== Normal patient ==========")
    for patient in patients:
        if patient["emergency"] == "no":
            print(f'Patient name: {patient["name"]}')
            print(f'Patient age: {patient["age"]}')
            print(f'Patient problem: {patient["problem"]}')
            print(f'Patient time: {patient["time"]}')


def delete_patient():
    patient_id = int(input("Enter patient id: "))

    for patient in patients:
        if patient["id"] == patient_id:
            patients.remove(patient)
            print("Patient deleted")
            return

    print("Patient not found")


def count_patient():
    emergency_count = 0
    normal_count = 0

    for patient in patients:
        if patient["emergency"] == "yes":
            emergency_count += 1
        else:
            normal_count += 1

    print("Emergency patients:", emergency_count)
    print("Normal patients:", normal_count)
    print("Total patients:", len(patients))


while True:
    print("1. Add patient")
    print("2. Search patient")
    print("3. Display appointment")
    print("4. Delete patient")
    print("5. Count patient")
    print("6. Exit")

    try:
        choice = int(input("Enter your choice: "))
    except ValueError:
        print("Invalid input. Please enter a number between 1 and 6.")
        continue

    if choice == 1:
        add_patient()
    elif choice == 2:
        search_patient()
    elif choice == 3:
        display_appointment()
    elif choice == 4:
        delete_patient()
    elif choice == 5:
        count_patient()
    elif choice == 6:
        break
    else:
        print("Invalid choice")
````
