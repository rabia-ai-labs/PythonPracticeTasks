# Python Practice Questions

1. Even or Odd
Take a number from the user and check whether it is even or odd.
2. Simple Calculator
Take two numbers and an operator (+, -, *, /) from the user. Perform the selected operation and display the result.
3. Student Marks
Take marks of 3 subjects from the user. Calculate the total and percentage, then display Pass if the percentage is 50 or above, otherwise display Fail.
4. Favorite Fruits
Create a list of 5 fruits. Ask the user to enter a fruit and check whether that fruit exists in the list or not.
5. Student Information
Create a dictionary containing a student's name, age, and favorite subject. Display all the information and use a condition to check whether the student's age is 18 or abo

```python
# Question: 1
number = int(input("Enter a number: "))

if number % 2 == 0:
  print ("Even")
else:
  print("Odd")
```

```python
# Question: 2
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -,*,/ )")

if operator == "+":
  print(num1 + num2)
elif operator == "-":
  print(num1 - num2)
elif operator == "*":
  print(num1 * num2)
elif operator == "/":
  print(num1 / num2)
```

```python
# Question 3
marks1 = float(input("Enter marks of subject 1:"))
marks2 = float(input("Enter marks of subject 2:"))
marks3 = float(input("Enter marks of subject 3:"))

total = marks1 + marks2 + marks3
percentage = total / 300 * 100

print("Total:", total)
print("Percentage:", percentage)

if percentage >= 50:
  print("Pass")
else:
  print("Fail")
```

```python
# Question 4
fruits = ["mango", "strawberry", "kiwi", "cherry", "peach"]

fruit = input("Enter fruit name: ")

if fruit in fruits:
  print("Fruits exist in the list")
else:
  print("Fruit does not exist in the list")
```

```python
# Question 5

student = {
    "name": "Rabia",
    "age": 18,
    "favourite_subject": "Computer Science"
}

print("Name:", student["name"])
print("Age:", student["age"])
print("Favourite subject:", student["favourite_subject"])

if student["age"] >= 18:
  print("Student is 18 or above")
else:
  print("Student is below 18")
```
